```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import mutual_info_classif
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("VehicleAccident.csv", low_memory=False)

# Target encoding
df["CRASH_CLASS"] = df["CRASH CLASSIFICATION DESCRIPTION"].map({
    "Non-Reportable": 0,
    "Property Damage Only": 1,
    "Personal Injury Crash": 2,
    "Fatality Crash": 3
})

print("Null counts in target:", df["CRASH_CLASS"].isnull().sum())

# Convert binary Y/N columns to 0/1
bin_cols = [
    "COLLISION ON PRIVATE PROPERTY",
    "PEDESTRIAN INVOLVED",
    "ALCOHOL INVOLVED",
    "DRUG INVOLVED",
    "SEATBELT USED",
    "MOTORCYCLE INVOLVED",
    "MOTORCYCLE HELMET USED",
    "BICYCLED INVOLVED",
    "BICYCLE HELMET USED",
    "WORK ZONE",
    "WORKERS PRESENT"
]

def encode_binary(x):
    return x.map({"Y":1,"N":0,"1":1,"2":1,"9":0})

for c in bin_cols:
    if c in df.columns:
        df[c+"_bin"] = encode_binary(df[c])

# Keep only numeric columns for now
numeric_df = df.select_dtypes(include=['int64','float64'])

# Pearson correlation with the target
corr = numeric_df.corr()["CRASH_CLASS"].sort_values(ascending=False)
print("\n===== PEARSON CORRELATION WITH TARGET =====\n")
print(corr)

# Save for me
corr.to_csv("correlation_with_target.csv")

# -------------------------
# Mutual Information (best for mixed data)
# -------------------------
# Prepare X, y
df_mi = df.copy()

# Fill NA
df_mi = df_mi.fillna("missing")

# Encode all categorical
df_encoded = df_mi.copy()
for col in df_mi.columns:
    if df_mi[col].dtype == "object":
        df_encoded[col] = LabelEncoder().fit_transform(df_mi[col].astype(str))

X = df_encoded.drop(columns=["CRASH_CLASS"])
y = df_encoded["CRASH_CLASS"]

mi_scores = mutual_info_classif(X, y, discrete_features='auto')
mi_series = pd.Series(mi_scores, index=X.columns).sort_values(ascending=False)

print("\n===== MUTUAL INFORMATION SCORES =====\n")
print(mi_series.head(20))
mi_series.to_csv("mutual_information_scores.csv")

# -------------------------
# Feature Importance using RandomForest
# -------------------------

rf = RandomForestClassifier(n_estimators=200)
rf.fit(X, y)

feat_importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)

print("\n===== RANDOM FOREST FEATURE IMPORTANCE =====\n")
print(feat_importance.head(20))
feat_importance.to_csv("feature_importance_rf.csv")

print("\nAll files saved: correlation_with_target.csv, mutual_information_scores.csv, feature_importance_rf.csv")
```
